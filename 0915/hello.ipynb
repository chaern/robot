```python
from pymcprotocol import Type3E
import time
PLC_IP = "192.168.3.100"
PLC_PORT = 1026
plc = Type3E()
try:
    # PLC 연결
    plc.connect(PLC_IP, PLC_PORT)
    print("PLC 연결 성공")
    message = "HELLO"
    ascii_data = [ord(c) for c in message]
    print("전송 문자열 :", message)
    print("ASCII DATA :", ascii_data)
    plc.batchwrite_wordunits( headdevice="D100", values=ascii_data   )
    print("Python → PLC : HELLO 전송 완료")
    check_data = plc.batchread_wordunits( headdevice="D100",  readsize=5   )
    print("D100~D104 확인 :", check_data)
    plc.batchwrite_bitunits(   headdevice="M100",   values=[1]    )
    print("M100 ON")
    while True:
        done = plc.batchread_bitunits(  headdevice="M101",   readsize=1  )
        print("M101 =", done)
        if done[0] == 1:
            break
        time.sleep(0.1)
    print("PLC 처리 완료")
    # =========================
    # D200 ~ D204 읽기
    # =========================
    receive_data = plc.batchread_wordunits( headdevice="D200", readsize=5   )
    print("PLC에서 받은 데이터 :", receive_data)
    # ASCII → 문자열
    receive_message = ''.join(  chr(value) for value in receive_data    )
    print("PLC → Python :", receive_message)
    # M100 OFF
    plc.batchwrite_bitunits(  headdevice="M100", values=[0]   )
    print("M100 OFF")
    time.sleep(0.2)
except Exception as e:
    print("통신 오류 :", e)
finally:
    try:
        plc.close()
        print("PLC 연결 종료")
    except:
        pass
```
